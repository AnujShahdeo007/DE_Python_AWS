# Example Custom Exception

```python
# Age Validation
class AgeError(Exception):
    pass

age=15
try:
    if age <18:
        raise AgeError("Age must be 18 or above")
    print("Eligible")
except AgeError as e:
    print("Error:",e)
```

```python
class InvalidMarksError(Exception):
    pass

marks=101
try:
    if marks < 0 or marks>100:
        raise InvalidMarksError("Marks mut be between 0 and 100") 
    print("Valid Marks")
except InvalidMarksError as e:
    print(e)
```

```python
class InsuffientBalanceError(Exception):
    def __init__(self, balance,amount):
        self.balance=balance
        self.amount=amount 
    def __str__(self):
        return f"Available balance is {self.balance}, but tried to withdraw {self.amount}"
```

```python
def withdraw(balance,amount):
    if amount>balance:
        raise InsuffientBalanceError (balance,amount)
    return balance-amount

try:
    print(withdraw(4000,8000))
except InsuffientBalanceError as e:
    print("Transection Failed:",e)
```
