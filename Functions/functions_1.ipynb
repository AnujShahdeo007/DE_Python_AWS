# Lambda

```python
add=lambda x,y:x+y
print(add(5,3))

#function(x,y)
#    return x+y
```

```python
square=lambda n:n*n
print(square(5))
```

```python
check=lambda x : "Even" if x%2==0 else "Odd"
print(check(10))
```

```python
nums=[1,2,3,4]
result= list(map(lambda x:x*10,nums))
print(result)
```

```python
check=lambda n : "Positive" if n>0 else "Negative" if n<0 else "Zero"

print(check(0))
```

```python
multiply=lambda x :(lambda y : x*y)
print(multiply(5)(3))
```

```python
cal= lambda x : lambda y : lambda z : x+y+z
print(cal(1)(2)(3))
```

# Recursion 

```python
def countdown(n):
    if n==0:
        return 
    print(n)
    countdown(n-1)
```

```python
countdown(6)
```

```python
def factorial(n):
    if n==1 or n==0:
        return 1
    return n * factorial(n-1)
```

```python
factorial(0)
```

```python
def total(n):
    if n==0:
        return 0
    return n + total(n-1)
```

```python
total(100)
```

```python
def count_digits(n):
    if n==0:
        return 0
    return 1+count_digits(n//10)
```

```python
print(count_digits(12345))
```

```python
def count_digit_loop(n):
    count=0
    while n>0:
        count +=1
        n=n//10
    return count 
```

```python
count_digit_loop(1234574848)
```
