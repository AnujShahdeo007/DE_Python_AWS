Create SparkSession

    SparkSession.builder
    .appName("DataFrameDemo")
    .master("local[*])
    .getOrCreate()
)

Step-3 Prepare Emp Data 

data=[
    (101,"A","IT",50000),

df=spark.createDataFrame(data,columns)
