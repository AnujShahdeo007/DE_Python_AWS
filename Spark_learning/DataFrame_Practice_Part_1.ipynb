```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Demo") \
    .getOrCreate()
```

```python
data=[
    (101,"A","IT",50000),
    (101,"A","IT",60000),
    (101,"A","IT",80000),
    (101,"A","IT",50000),
    (101,"A","IT",50000),

]

columns=["emp_id","name","department","salary"]

df=spark.createDataFrame(data,columns)
```

```python
df.printSchema()
```

```python
from pyspark.sql.types import (
    StringType,
    StructField,
    IntegerType,
    DoubleType,
    StructType
)
data=[
    (101,"A","IT",50000), # int 
    (101,"A","IT",60000), # int 
    (101,"A","IT",80000.0), # float 
    (101,"A","IT",50000), # int
    (101,"A","IT",50000.0), # float 


]

schema= StructType([
    StructField("emp_id",IntegerType(),False),
    StructField("name",StringType(),True),
    StructField("Department",StringType(),True),
    StructField("salary",DoubleType(),True)
])

# convert salary to float 
cleaned_data=[
    (emp_id,name,dept,float(salary))
    for emp_id,name,dept,salary in data
]

df_schema=spark.createDataFrame(cleaned_data,schema=schema)
df_schema.show()
df_schema.printSchema()
```
