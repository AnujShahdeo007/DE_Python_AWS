Prob-1 : Student Marks Mangement System 

You are building a Student marks management System for institute 

Each student has :
    - Student id 
    - name 
    - Marks in subjects 

students={
    101:{"name":"Priyanka","maths":85,"science":90,"english":75},
    102:{"name":"Nilam","maths":88,"science":60,"english":85},
    103:{"name":"Priya","maths":30,"science":0,"english":5}
}

1. Print all student names 
2. Print all student Ids 
3. Print Full student details 
4. Add new Student 
5. Update marks of a student 
    "Priyanka" maths= 100

6. Calculate total marks of each student 
7. Find the topper of the class 
8. Remove a student who left the institute # Suppose Stiudent ID 103 left 
9. Check if a student exists #102
10. Sort students by total marks 


```python
students={
    101:{"name":"Priyanka","maths":85,"science":90,"english":75},
    102:{"name":"Nilam","maths":88,"science":60,"english":85},
    103:{"name":"Priya","maths":30,"science":0,"english":5}
}
```

```python
#1. 
for student_id,details in students.items():
    print(details["name"])
```

```python
#2. 
for student_id in students.keys():
    print(student_id)
```

```python
#3. 
for student_id,details in students.items():
    print("Student ID:",student_id)
    print("Details:",details)
    print()
```

```python
#4.
students[104]={
    "name":"Rahul",
    "maths":95,
    "science":92,
    "english":89
}
print(students)
```

```python
#5. 
for student_id,details in students.items():
    if details["name"]=="Priyanka":
        details["maths"]=100
        break
print(students)
```

```python
#6. 
for student_id,details in students.items():
    total=details["maths"]+details["science"]+details["english"]
    print(details["name"],"---> Total Marks=",total)
```

```python
topper_name=""
highest_marks=-1 #265 #276
for student_id,details in students.items():
    total=details["maths"]+details["science"]+details["english"]
    if total>highest_marks:
        highest_marks=total
        topper_name=details["name"]
print("Topper:",topper_name)
print("highest_marks:",highest_marks)
```

```python
removed_students=students.pop(103,None)
print("Removed:",removed_students)
print(students)
```

```python
studnet_id=102
if studnet_id in students:
    print("Student Exists")
else:
    print("Student does not exists")
```

```python
sorted_students=sorted(
    students.items(),
    key=lambda item:item[1]["maths"]+item[1]["science"]+item[1]["english"],
    reverse=True
)

for studnet_id,details in sorted_students:
    total=details["maths"]+details["science"]+details["english"]
    print(student_id,details["name"],"-->",total)
```
