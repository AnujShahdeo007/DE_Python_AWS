Senario :
    Student Performance Analyzer 

A institute wants to analyze students marks from multiple tests

Each student record contains:
- student name 
- marks in 3 subjects 
- attendence status 
- fee payment status 


Institute wants to :
- count the total no of students 
- find the highest and lowest score 
- calculate total marks 
- sort students by performance 
- Reverse ranking 
- check if any student failed 
- check if all students paid fees 
- convert raw data into list
- generate rank number 
- combine name and marks 
- creates a roll no 
- transform marks 
- filter passed students 
- validate the data type
- inspect object identity and type 
- print a clean report 

```python
students=[
    ["Vamshi",[78,85,90],True,True],
    ["Ravi",[45,52,40],True,False],
    ["Neha",[88,91,95],True,True],
    ["Simran",[35,60,55],True,True],
    ["Aman",[67,72,55],False,True],
    ["Gaurav",[8,5,0],True,True]
    
]
```

```python
print("==================================STUDENT PERFORMANCE REPORT============================================")
#. Extract names into a new list
names= list(student[0] for student in students)
print("Student_names:",names)
#Q 1. count the total no of students 
print("total Students:",len(students))

# sum() : total marks for each students 
studnets_total=[(student[0],sum(student[1]))for student in students]
print("Student_total_Marks:",studnets_total)

# MAX
topper=max(studnets_total,key=lambda x:x[1])

print("Topper:",topper)
```
