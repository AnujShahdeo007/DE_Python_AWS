BITWISE OPERTOR

(& | ^ ~ << >>)

Bitwise operator works on binary (bits) not on deciaml values---> Python converts numbers to binary to perform opearations, then convert back to decimal 

Decimal                 Binary 
0                       0000
1                       0001
2                       0010
3                       0011
4                       0100
5                       0101
6                       0110
7                       0111



```python
print(5 & 3)
# Rule Bit is 1 only if both bits are 1 
# 5 = 0101
# 3 = 0011
# --------
#     0001 #--> 1
```

```python
print( 5|3)
# Rule Bit is 1 if any bit 1
# 5 = 0101
# 3 = 0011
# --------
#     01111 #--> 7
```

```python
print(5 ^ 3)
# Rule Bit 1 only if bits are diffrent
# 5 = 0101
# 3 = 0011
# --------
#     0110 ---> 6
```

```python
print(~5) 
# Rule: Flips all bits and add minus one 

# ~x= -(x+1)

# ~5 = -(5+1) = -6
```

```python
print(~(-3))
```

```python
print(5<<1)
#Rule; Shift bits left ---> Multiply by 2 
# 5 = 0101
#Shift left 1 
#-> 1010
```

```python
print(20>>2)
# Rule: Shift bits right --> Divide by 2 (floor)
# 20/2**2
```
